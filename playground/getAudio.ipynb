```python
from datasets import load_dataset
import torchaudio

dataset = load_dataset("AI4A-lab/RecruitView")
```

```python
video_file = dataset["train"][0]["video"]

# Get the path from _hf_encoded
video_path = video_file._hf_encoded['path']
print(video_path)  # /Users/rayyanzaid/.cache/hu...
```

```python
import ffmpeg
import numpy as np

video_path = video_file._hf_encoded['path']

out, _ = (
    ffmpeg
    .input(video_path)
    .output('pipe:', format='f32le', acodec='pcm_f32le', ac=1, ar='16000')
    .run(capture_stdout=True, capture_stderr=True)
)

audio_array = np.frombuffer(out, np.float32)
print(audio_array.shape)
```
